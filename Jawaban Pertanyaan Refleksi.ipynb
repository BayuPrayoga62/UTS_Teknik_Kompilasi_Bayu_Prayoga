1. Mengapa fungsi `power()` harus dipanggil di dalam `term()`, bukan sebaliknya? Jelaskan kaitannya dengan *Operator Precedence*.
Dalam *Recursive Descent Parser*, urutan pemanggilan fungsi menentukan prioritas operator (*Operator Precedence*). Aturannya adalah: operator dengan **prioritas tertinggi** akan diletakkan di fungsi yang **paling dalam** (paling terakhir dipanggil). Karena pemangkatan `(^)` memiliki prioritas **lebih tinggi daripada perkalian/pembagian (*, /)**, maka fungsi `term()` yang menangani `* dan /` harus mendelegasikan pengecekan ke fungsi `power()` terlebih dahulu. Ini memastikan bahwa operasi pangkat akan diselesaikan dan dijadikan sebuah "node" sebelum dikalikan atau dibagi.

2. Apa yang terjadi pada fase **Analisis Semantik** jika variabel `z` digunakan dalam kode sumber tetapi tidak ada di `symbol_table`?
Proses parsing akan terhenti dan program akan melemparkan pengecualian (*Exception*). Hal ini bisa dilihat pada fungsi `factor()`, di mana terdapat pengecekan `if token not in self._env:`. 3. Jelaskan mengapa dalam TAC, instruksi untuk `a ^ 2` harus muncul sebelum instruksi untuk `+`.
TAC *(Three Address Code)* dibuat berdasarkan struktur *Abstract Syntax Tree* (AST) dengan metode *Bottom-Up* atau *Post-Order Traversal* (anak-anak node dikerjakan terlebih dahulu sebelum *parent*-nya). Karena operator `^` memiliki prioritas lebih tinggi, letak node `^` di dalam pohon (AST) akan lebih bawah dibandingkan node `+`. Oleh karena itu, operasi `a ^ 2` harus dieksekusi terlebih dahulu dan hasilnya disimpan di variabel *temporary* (misalnya `t1`), barulah nilai `t1` tersebut ditambahkan pada operasi penjumlahan (`+`) yang posisinya berada di atasnya.
